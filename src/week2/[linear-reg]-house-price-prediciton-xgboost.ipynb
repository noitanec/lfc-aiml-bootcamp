```python
# This Python 3 environment comes with many helpful analytics libraries installed

import numpy as np # linear algebra
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)
from sklearn.preprocessing import LabelEncoder
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler
```

Dataset link: https://www.kaggle.com/competitions/neolen-house-price-prediction/data

```python
df = pd.read_csv("../data/house-price-prediction/train.csv")
# test_df = pd.read_csv("../data/house-price-prediction/test.csv")
df.shape
```

```python
feature_col = ['LotArea', 'TotalBsmtSF', 'PoolArea']
tgt_col  = 'SalePrice'
```

```python
# Dropping 'Id' column as it has no use in training
df = df.drop(['Id'], axis=1)
df = df.drop(['Id'], axis=1)
```

```python
print(f"Shape of df: {df.shape}")
```

```python
from sklearn.model_selection import train_test_split

train, test = train_test_split(df, test_size=.3)
train.shape, test.shape
```

```python
train_x, train_y = train[feature_col], train[tgt_col]
test_x, test_y = test[feature_col], test[tgt_col]
```

```python
train_x.shape, train_y.shape, test_x.shape, test_y.shape
```

## Linear Regression model 

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(train_x,train_y)
```

```python
model.coef_, model.intercept_
```

```python
y_pred = model.predict(test_x)
```

```python
y_pred.shape
```

```python
np.sum(np.sqrt((test_y.to_numpy() - y_pred)**2))
```
